# Занятие 4, демо 3. Шесть строк

Инструмент, с которым вы уйдёте с занятия. Встретив незнакомую генеративную
модель, заполните по её описанию 6 строк:

| # | Строка |
|---|---|
| 1 | откуда стартуем |
| 2 | какой путь |
| 3 | какая связка |
| 4 | что минимизируем |
| 5 | каким солвером идём |
| 6 | сколько вызовов тратим |

Незаполненная строка - это тоже результат: она показывает, чего в описании
не хватает.

## Короткое 1

> Модель порождает картинку, стартуя из гауссова шума. Промежуточную точку берут на отрезке между шумом и картинкой, с весами по времени. Обучена по квадрату отклонения от разности концов. При генерации 20 равномерных шагов Эйлера.

## Короткое 2

> Та же конструкция, но после первого обучения сделали переспрямление и обучили поле заново. Остальные настройки обучения и сэмплера сохранили, и теперь хватает 2 вызовов.

## Короткое 3

> Наш сэмплер даёт то же качество за 10 шагов вместо 50.

## Развёрнутое 1

> Модель порождает звук. Старт - гауссов шум той же длины, что и результат. Путь между шумом и записью задан постепенным зашумлением записи по расписанию. Сеть учится предсказывать добавленный шум. При генерации 50 равномерных шагов детерминированного солвера, по одному вызову на шаг.

## Развёрнутое 2

> Задача та же, что у нас сегодня. Но пары шум-объект перед обучением подбирают: берут пачку из 256 шумов и 256 объектов, составляют 256 пар, используя каждый шум и каждый объект ровно один раз, и минимизируют суммарную длину отрезков. Дальше всё как обычно, 20 шагов Эйлера.

## Развёрнутое 3

> Модель выдаёт картинку за 2 вызова сети. Обучена на прямых отрезках, потом дистиллирована.

## Развёрнутое 4

> Модель работает не с картинкой напрямую, а со сжатым представлением: сначала картинку сжимают обученным кодировщиком в небольшой набор чисел, генерируют этот набор, потом разворачивают обратно. Старт - гауссов шум в этом сжатом пространстве, пары независимые, обучающий путь - отрезок, учат по квадрату отклонения от разности концов. При генерации 30 шагов.

## Развёрнутое 5

> Новый метод генерирует за 4 шага вместо 100, ускорение в 25 раз.
